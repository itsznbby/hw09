```python
測試一:
# ==================================================
# HW09「兩場活動報名整理器」- 標準格式輸出版
# ==================================================

# 1. 固定的兩個活動名稱，以 tuple 保存。
activity_names = ("桌遊體驗", "電影欣賞")

# 2. 收到的原始文字名單，請保留原值。
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# 3. 用活動名稱查詢每人費用，以 dict 保存。
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# --------------------------------------------------
# 步驟一：資料清理（清理大小寫、去除首尾半形空白）
# --------------------------------------------------
board_clean_str = board_raw.strip().lower()
movie_clean_str = movie_raw.strip().lower()

# --------------------------------------------------
# 步驟二：拆分字串為 List（保留重複）
# --------------------------------------------------
board_list = board_clean_str.split(", ")
movie_list = movie_clean_str.split(", ")

# --------------------------------------------------
# 步驟三：轉為 Set（去重與集合運算）
# --------------------------------------------------
board_set = set(board_list)
movie_set = set(movie_list)

# 進行集合運算：
union_set = board_set | movie_set          # 聯集：全部參加者
intersection_set = board_set & movie_set   # 交集：共同參加者
board_only_set = board_set - movie_set     # 差集：只參加桌遊者
movie_only_set = movie_set - board_set     # 差集：只參加電影者

# 將集合轉換為已排序的 list，方便美觀地呈現
board_unique_list = sorted(list(board_set))
movie_unique_list = sorted(list(movie_set))
common_list = sorted(list(intersection_set))
board_only_list = sorted(list(board_only_set))
movie_only_list = sorted(list(movie_only_set))
all_participants_list = sorted(list(union_set))

# --------------------------------------------------
# 步驟四：計算人數、人次與切片
# --------------------------------------------------
# 報名人次：兩場原始 list 長度相加
total_registrations = len(board_list) + len(movie_list)

# 擷取全部參加者（不重複）的前兩位作為通知名單
notification_list = all_participants_list[:2]

# --------------------------------------------------
# 步驟五：計算收費（去重後人數 * 單價，轉為 float 以符合 .2f 格式化）
# --------------------------------------------------
board_fee_float = float(unit_fees["桌遊體驗"])
movie_fee_float = float(unit_fees["電影欣賞"])

board_total_fee = len(board_set) * board_fee_float
movie_total_fee = len(movie_set) * movie_fee_float
total_revenue = board_total_fee + movie_total_fee

# --------------------------------------------------
# 步驟六：格式化輸出（符合指定參考格式）
# --------------------------------------------------
print("HW09 活動報名摘要")
print(f"活動：{activity_names[0]}，每人 {board_fee_float:.2f} 元")
print(f"活動：{activity_names[1]}，每人 {movie_fee_float:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_unique_list}")
print(f"電影去重名單：{movie_unique_list}")
print(f"桌遊人數：{len(board_set)} 人")
print(f"電影人數：{len(movie_set)} 人")
print(f"共同參加者：{common_list}")
print(f"共同人數：{len(intersection_set)} 人")
print(f"只參加桌遊者：{board_only_list}")
print(f"只參加電影者：{movie_only_list}")
print(f"全部參加者：{all_participants_list}")
print(f"總不重複人數：{len(union_set)} 人")
print(f"報名人次：{total_registrations} 人次")
print(f"通知名單前兩位：{notification_list}")
print(f"桌遊應收：{board_total_fee:.2f} 元")
print(f"電影應收：{movie_total_fee:.2f} 元")
print(f"總應收：{total_revenue:.2f} 元")
```
